```python
import numpy as np 
import random
```

experiment 1: pick between 2 levers

```python
mu1 = 1
sig1 = 0.5 
mu2 = 0.9 
sig2 = 0.5 

num_tries = 10000

def bandit_experiment(prob):
    
    total_value = 0

    est1 = 0.0 
    est2 = 0.0
    tries1 = 0 
    tries2 = 0
    total1 = 0
    total2 = 0 
    
    for i in range(num_tries): 
        pick = random.uniform(0, 1) # 70% chance to exploit
        
        if (est1 > est2 and pick < prob) or (est1 < est2 and pick >= prob): 
            val = np.random.normal(mu1, sig1)
            total_value += val
            tries1 += 1
            total1 += val 
            est1 = total1 / tries1
    
        else:
            val = np.random.normal(mu2, sig2)
            total_value += val
            tries2 += 1
            total2 += val
            est2 = total2 / tries2 

    return total_value, est1, est2

print(bandit_experiment(0.5))
print(bandit_experiment(0.6))
print(bandit_experiment(0.7))
print(bandit_experiment(0.8))
print(bandit_experiment(0.9))
print(bandit_experiment(0.95))
```
